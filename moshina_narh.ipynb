```python
moshina_narh.columns = ['Turi', 'Yil', 'Sotish narhi', 'Yurgan kmm', 'Yonilgi turi', 'Sotuv turi', 'Karobka', 'Egasi']
moshina_narh.head()
```

```python
moshina_narh.sort_index(inplace=True)
moshina_narh.head()
```

```python
moshina_narh['Sotish narhi'].max()
```

```python
plt.figure(figsize=(15, 10))
sns.scatterplot(x='Yil', y='Sotish narhi', data=moshina_narh)
plt.show()
# bizda Yil va Sotish narhi orasidagi grafikni chiqarib oldik bunda biz 2000 Yildan kngi malumotlarni chiqarib filtr qilib olamiz kn gi kodda
```

```python
moshina_narh=moshina_narh[moshina_narh.Yil>2000] # aitganimizdak filtrlab oldik
plt.figure(figsize=(15, 10))
sns.scatterplot(x='Yil', y='Sotish narhi', color='r', data=moshina_narh)
plt.show()
```

```python
plt.figure(figsize=(15, 10))
sns.scatterplot(x='Yurgan kmm', y='Sotish narhi', data=moshina_narh)
plt.show()
```
