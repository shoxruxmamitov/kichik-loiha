```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
!wget clone https://raw.githubusercontent.com/muhammadjon-merzaqulov/Car_Sales_Dashboard/refs/heads/main/BMW_Sales_Data.csv
```

```python
statistika = pd.read_csv('BMW_Sales_Data.csv', index_col=0)
statistika.head()
```

```python
statistika.shape
```

```python
statistika.size
```

```python
statistika.isnull().sum()
# mavjud bolmagan qimatlar sonini chiqarib olamiz
```

```python
statistika = statistika.dropna()
statistika.head()
# nan qimatlarni olib tashladik
```

```python
statistika.isnull().sum()
# mavjud bolmagan qimatlar soni 0 ta
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(x = 'Year', y = 'Revenue', data=statistika)

plt.title('Revenue by Year')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(x = 'Year', y = 'Revenue', hue='Region', data=statistika)

plt.title('Revenue by Year')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.grid()
plt.show()
```

```python
statistika.to_csv('Sales_car.csv', index=False)
```
